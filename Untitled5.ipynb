1. Project Overview

Title: Exploratory Data Analysis of the Ames Housing Dataset

Dataset: House Prices: Advanced Regression Techniques (Kaggle), file train.csv, approximately 1,460 rows and 81 columns.

Objective: Use exploratory data analysis only (no machine learning) to understand which factors drive house sale prices and practise the complete EDA workflow.

Step 1: Setup and First Look

Goal: Load the data and understand its shape, types, and missing values.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

df = pd.read_csv("train.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[missing_summary["Missing Values"] > 0]

display(missing_summary.sort_values(
    by="Missing Percentage", ascending=False
))
```

Step 2: Classify Variables

Goal: Split columns into numerical, categorical, and ordinal groups.

```python
numeric_cols = df.select_dtypes(include=['number']).columns.tolist()

categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

ordinal_map = {
    "ExterQual": ["Po", "Fa", "TA", "Gd", "Ex"],
    "KitchenQual": ["Po", "Fa", "TA", "Gd", "Ex"]
}

for column, categories in ordinal_map.items():
    df[column] = pd.Categorical(
        df[column],
        categories=categories,
        ordered=True
    )

summary_data = []

for col in df.columns:

    if col in ordinal_map or col == "OverallQual":
        variable_type = "Ordinal"

    elif col in numeric_cols:
        variable_type = "Numerical"

    else:
        variable_type = "Categorical"

    summary_data.append({
        "Column": col,
        "Type": variable_type,
        "Unique Values": df[col].nunique(),
        "Example Values": df[col].dropna().unique()[:5]
    })

summary_df = pd.DataFrame(summary_data)

display(summary_df)
```

Step 3: Descriptive Statistics

Goal: Calculate mean, median, mode, standard deviation, variance, range, IQR, skewness and kurtosis.


```python
display(df.describe())

for column in ["SalePrice", "GrLivArea", "LotArea"]:

    print("\nStatistics for:", column)

    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode()[0])
    print("Standard Deviation:", df[column].std())
    print("Variance:", df[column].var())

    print("Range:", df[column].max() - df[column].min())

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    print("IQR:", q3 - q1)
    print("Skewness:", df[column].skew())
    print("Kurtosis:", df[column].kurt())

    if df[column].mean() > df[column].median():
        print("Distribution: Right Skewed")

    elif df[column].mean() < df[column].median():
        print("Distribution: Left Skewed")

    else:
        print("Distribution: Symmetrical")

print("\nNeighborhood Value Counts:")
display(df["Neighborhood"].value_counts())

print("\nHouse Style Value Counts:")
display(df["HouseStyle"].value_counts())
```

Step 5: Correlation and Covariance

Goal: Identify numerical features that have the strongest relationship with SalePrice.

```python
numeric_data = df.select_dtypes(include=np.number)

cov_matrix = numeric_data.cov()

corr_matrix = numeric_data.corr()

print("Covariance Matrix:")
display(cov_matrix)

print("\nCorrelation Matrix:")
display(corr_matrix)

top_corr = corr_matrix["SalePrice"].drop(
    "SalePrice"
).abs().sort_values(ascending=False)

print("\nTop 10 Features Correlated with SalePrice:")
display(top_corr.head(10))

print("\nTop 3 Features:")

for column in top_corr.head(3).index:

    print("\nFeature:", column)

    print(
        "Correlation:",
        corr_matrix.loc[column, "SalePrice"]
    )

    print(
        "Covariance:",
        cov_matrix.loc[column, "SalePrice"]
    )
```

Step 8: Outlier Detection

Goal: Detect outliers using two methods: IQR and Z-score. Compare their results.

```python
def iqr_outliers(series):

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    return series[(series < lower) | (series > upper)]
```

```python
def zscore_outliers(series, threshold=3):

    z = (series - series.mean()) / series.std()

    return series[z.abs() > threshold]
```

```python
for column in ["GrLivArea", "SalePrice"]:

    iqr_result = iqr_outliers(df[column])

    z_result = zscore_outliers(df[column])

    print("\nColumn:", column)

    print("IQR Outliers:", len(iqr_result))

    print("Z-score Outliers:", len(z_result))

    overlap = set(iqr_result.index) & set(z_result.index)

    print("Overlapping Outliers:", len(overlap))
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["GrLivArea"])

plt.title("Outliers in Living Area")

plt.show()
```

```python
living_outliers = iqr_outliers(df["GrLivArea"])

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="GrLivArea",
    y="SalePrice"
)

plt.scatter(
    df.loc[living_outliers.index, "GrLivArea"],
    df.loc[living_outliers.index, "SalePrice"],
    color="red",
    label="Outliers"
)

plt.title("Living Area vs SalePrice with Outliers")

plt.legend()

plt.show()
```

Step 9: Data Quality Checks

Goal: Identify duplicate records, inconsistent categories and missing values. Clean the dataset.

```python
df_clean = df.copy()

duplicate_count = df_clean.duplicated().sum()

print("Duplicate Rows:", duplicate_count)

df_clean = df_clean.drop_duplicates()

print("New Dataset Shape:", df_clean.shape)
```

```python
for column in df_clean.select_dtypes(include="object").columns:

    df_clean[column] = df_clean[column].str.strip()

print("Categorical values standardized.")
```

```python
none_columns = [
    "PoolQC",
    "Alley",
    "Fence",
    "FireplaceQu",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",
    "BsmtQual",
    "BsmtCond",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2",
    "MasVnrType"
]

for column in none_columns:

    if column in df_clean.columns:

        df_clean[column] = df_clean[column].fillna("None")

# Fill LotFrontage using Neighborhood median

df_clean["LotFrontage"] = df_clean.groupby(
    "Neighborhood"
)["LotFrontage"].transform(
    lambda x: x.fillna(x.median())
)

df_clean["LotFrontage"] = df_clean["LotFrontage"].fillna(
    df_clean["LotFrontage"].median()
)

# Fill remaining numerical missing values

for column in df_clean.select_dtypes(include=np.number).columns:

    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )

# Fill remaining categorical missing values

for column in df_clean.select_dtypes(include="object").columns:

    df_clean[column] = df_clean[column].fillna("Unknown")

print("Remaining Missing Values:")

print(df_clean.isnull().sum().sum())

print("\nCleaned Dataset Shape:")

print(df_clean.shape)
```

```python
print(
    "YearBuilt greater than YrSold:",
    (df_clean["YearBuilt"] > df_clean["YrSold"]).sum()
)

for column in ["GrLivArea", "LotArea", "TotalBsmtSF", "GarageArea"]:

    print(
        "Negative values in",
        column,
        ":",
        (df_clean[column] < 0).sum()
    )
```
